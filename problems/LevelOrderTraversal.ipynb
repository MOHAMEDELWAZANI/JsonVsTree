# 102. Binary Tree Level Order Traversal

**Medium**

Given the `root` of a binary tree, return *the level order traversal of its
nodes' values*. (i.e., from left to right, level by level).

---

**Example 1:**

```
        3
       / \
      9   20
         /  \
        15   7

Input:  root = [3, 9, 20, null, null, 15, 7]
Output: [[3], [9, 20], [15, 7]]
```

**Example 2:**

```
Input:  root = [1]
Output: [[1]]
```

**Example 3:**

```
Input:  root = []
Output: []
```

---

**Constraints:**

- The number of nodes in the tree is in the range `[0, 2000]`.
- `-1000 <= Node.val <= 1000`


### Where this comes from

```
while queue not empty:
```

Two lines change. Ask yourself:

- inside the inner loop, where does `node.val` go?
- after the inner loop finishes, what do you do with the group you collected?

Note the output is a **list of lists**, not a flat list - the shape of the
answer is the shape of the levels.

````python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
````

````python
from collections import deque

class Solution:
    def levelOrder(self, root: TreeNode) -> list[list[int]]:
        if root is None: return []
        q = deque()
        q.append(root)
        result = []
        while q:
            level = []
            size = len(q)
            for _ in range(size):
                i:TreeNode = q.popleft()
                level.append(i.val)
                if i.left: q.append(i.left)
                if i.right: q.append(i.right)
            result.append(level)
        return result
````

````python
# tests
sol = Solution()

# Example 1
t1 = TreeNode(3,
              TreeNode(9),
              TreeNode(20, TreeNode(15), TreeNode(7)))
print(sol.levelOrder(t1))     # [[3], [9, 20], [15, 7]]

# Example 2
print(sol.levelOrder(TreeNode(1)))    # [[1]]

# Example 3
print(sol.levelOrder(None))           # []

# perfect tree
t4 = TreeNode(1,
              TreeNode(2, TreeNode(4), TreeNode(5)),
              TreeNode(3, TreeNode(6), TreeNode(7)))
print(sol.levelOrder(t4))     # [[1], [2, 3], [4, 5, 6, 7]]

# lopsided - the levels have different widths
#        1
#       / \
#      2   3
#     / \   \
#    4   5   6
#             \
#              7
#               \
#                8
t8 = TreeNode(1,
              TreeNode(2, TreeNode(4), TreeNode(5)),
              TreeNode(3, None,
                TreeNode(6, None,
                  TreeNode(7, None,
                    TreeNode(8)))))
print(sol.levelOrder(t8))     # [[1], [2, 3], [4, 5, 6], [7], [8]]

# a chain - every level holds exactly one node
t5 = TreeNode(1, TreeNode(2, TreeNode(3, TreeNode(4, TreeNode(5)))))
print(sol.levelOrder(t5))     # [[1], [2], [3], [4], [5]]
````
