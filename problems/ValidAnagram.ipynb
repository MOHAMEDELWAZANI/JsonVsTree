### Think before you type

Two words are anagrams when they contain **the same letters with the same
counts**. `"aab"` and `"aba"` are anagrams; `"aab"` and `"abb"` are not - same
letters, wrong counts.

Three routes, from blunt to sharp:

1. **Sort both, compare.** `sorted("nagaram") == sorted("anagram")`? One line.
   What does sorting cost - and is that the bottleneck? `O(n log n)`.
2. **Count letters.** Build a tally per letter for `s`, do the same for `t`,
   compare the two tallies. What structure holds "thing -> count"? You used it in
   Two Sum. `O(n)`.
3. **One dict, walk twice.** Add for every letter of `s`, subtract for every
   letter of `t`. If the strings match, every count lands back at `0`. What has
   to be true about the leftover counts at the end?
