# 226. Invert Binary Tree

**Easy**

Given the `root` of a binary tree, invert the tree, and return *its root*.

Inverting a tree means turning it into its **mirror image**: every left child
becomes the right child and vice versa, all the way down.

---

**Example 1:**

```
     4                     4
   /   \                 /   \
  2     7      ->        7     2
 / \   / \             / \   / \
1   3 6   9            9   6 3   1

Input:  root = [4, 2, 7, 1, 3, 6, 9]
Output: [4, 7, 2, 9, 6, 3, 1]
```

**Example 2:**

```
Input:  root = [2, 1, 3]
Output: [2, 3, 1]
```

**Example 3:**

```
Input:  root = []
Output: []
```

---

**Constraints:**

- The number of nodes in the tree is in the range `[0, 100]`.
- `-100 <= Node.val <= 100`


### This one is built for recursion

You have leaned on your BFS queue for every tree problem so far. This is the one
where recursion is genuinely the shorter, clearer tool - so do it that way first.

Ask what inverting **one** node means, ignoring the rest of the tree:

> swap this node's `left` and `right`.

That is the whole operation - for a single node. But its children are still in
their old arrangement. So after you swap at the root, what must happen to the two
subtrees hanging off it? The same thing. That "same thing, on a smaller piece"
is the recursion.

- **Base case:** what is the mirror of an empty tree (`None`)? Return it.
- **Recursive step:** swap the two children, then invert each subtree.

One thing to test on yourself: does it matter whether you *swap first then
recurse*, or *recurse first then swap*? Trace Example 1 both ways on paper. (It
works either way - understanding *why* is the lesson.)

In Python you can swap without a temp variable: `a, b = b, a`. Use it.

---

**Then, if you want:** redo it with your BFS queue. Dequeue a node, swap its two
children, enqueue whichever children exist. Same answer, iterative shape - and a
good check that you understand what the recursion was really doing.

### A helper to *see* the result

Comparing trees by eye is hard. This prints a tree in level order (your #102
output) so you can confirm the mirror actually happened. Run this cell; don't
edit it.

````python
from collections import deque


def level_order(root):
    """Return the tree level by level - so you can see the shape."""
    if root is None:
        return []
    q, out = deque([root]), []
    while q:
        level = []
        for _ in range(len(q)):
            node = q.popleft()
            level.append(node.val)
            if node.left:  q.append(node.left)
            if node.right: q.append(node.right)
        out.append(level)
    return out
````

### A quieter observation

Invert a tree twice and you are back where you started - inverting is its own
undo. And notice the base case did the heavy lifting: the moment you correctly
say "the mirror of `None` is `None`", every leaf handles itself for free, because
a leaf's children *are* `None`. Most tree recursions live or die on the base
case, not the clever part.
