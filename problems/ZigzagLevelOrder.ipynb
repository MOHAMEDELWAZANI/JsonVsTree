# 103. Binary Tree Zigzag Level Order Traversal

**Medium**

Given the `root` of a binary tree, return *the zigzag level order traversal of
its nodes' values*. (i.e., from left to right, then right to left for the next
level and alternate between).

---

**Example 1:**

```
        3
       / \
      9   20
         /  \
        15   7

Input:  root = [3, 9, 20, null, null, 15, 7]
Output: [[3], [20, 9], [15, 7]]
```

**Example 2:**

```
Input:  root = [1]
Output: [[1]]
```

**Example 3:**

```
Input:  root = []
Output: []
```

---

**Constraints:**

- The number of nodes in the tree is in the range `[0, 2000]`.
- `-100 <= Node.val <= 100`


### The problem you accidentally named

Your BFS function was called `ZigZag` but printed plain level order. This is the
real thing.

Compare the two outputs on Example 1:

| | level 1 | level 2 | level 3 |
|---|---|---|---|
| #102 level order | `[3]` | `[9, 20]` | `[15, 7]` |
| #103 zigzag | `[3]` | `[20, 9]` | `[15, 7]` |

Only the direction of every **second** level flips. Levels 1, 3, 5... stay left
to right; levels 2, 4, 6... go right to left.

**Do not change how you traverse.** The queue must still take children left then
right, every level, always - flip the traversal and the levels come out wrong.
What changes is only what you do with a batch once you have collected it.

---

### Two ways to do it - and they are not equally good

**Way A** - collect the batch left to right as in #102, then reverse it when the
level index is odd.

**Way B** - keep a flag, and while collecting, put each value at the *front* of
the level list instead of the back when the flag says so.

Question before you pick: inserting at the front of a Python **list**
(`lst.insert(0, x)`) shifts every element that is already there. If a level
holds `k` nodes, what does building it cost? And what does `lst.reverse()` cost
for the same level?

One of them is `O(k)` per level, the other is `O(k^2)`. Work out which, then
choose. (And: is there a container where inserting at the front is `O(1)`? You
have already used it.)

````python
from zope.interface.common import optional


# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
````

### Your queue

Same as before - paste your `QueueNode` / `Queue`, or use `deque`.

````python
from collections import deque
from typing import Optional

class Node:
    def __init__(self , value:int = 0 , next=None):
        self.value = value
        self.next = next
class LinkedList:
    def __init__(self,head:Node = None,dale:Node=None) -> None:
        self.head = head
        self.dale = dale
        self.size = 0
        self.lst = []

    def printList(self):
        temp = self.head
        while temp:
            print(temp.value, end=" -> ")
            temp = temp.next
        print("None")

    def addFirst(self,value):
        self.size += 1
        i = Node(value)
        i.next = self.head
        self.head = i
        if self.dale is None :
            self.dale = i
        #self.head = Node(value, self.head)
    def addLast(self,value):
        self.size += 1
        i = Node(value)
        if self.dale:
            self.dale.next = i
        self.dale = i
        if self.head is None :
            self.head = i
    def append(self,value:int,position:str, index: Optional[int] = None):
        if position == "first":
            self.addFirst(value)
        else :
            self.addLast(value)
    def getList(self)->list:
            temp = self.head
            while temp:
                self.lst.append(temp.value)
                temp = temp.next
            return self.lst

````

````python
class Solution:
    def zigzagLevelOrder(self, root: TreeNode) -> list[list[int]]:
        if root is None: return []
        q = deque()
        q.append(root)
        result = []
        batch: int = 0
        while q:
            level:LinkedList = LinkedList()
            size = len(q)
            for _ in range(size):
                i:TreeNode = q.popleft()
                level.append(i.val,"first" if batch % 2 != 0 else "last")
                if i.left: q.append(i.left)
                if i.right: q.append(i.right)
            result.append(level.getList())
            batch +=1
        return result
````

````python
# tests
sol = Solution()

# Example 1
t1 = TreeNode(3,
              TreeNode(9),
              TreeNode(20, TreeNode(15), TreeNode(7)))
print(sol.zigzagLevelOrder(t1))     # [[3], [20, 9], [15, 7]]

# Example 2
print(sol.zigzagLevelOrder(TreeNode(1)))    # [[1]]

# Example 3
print(sol.zigzagLevelOrder(None))           # []

# perfect tree - level 3 is back to left-to-right
t4 = TreeNode(1,
              TreeNode(2, TreeNode(4), TreeNode(5)),
              TreeNode(3, TreeNode(6), TreeNode(7)))
print(sol.zigzagLevelOrder(t4))     # [[1], [3, 2], [4, 5, 6, 7]]

# lopsided - 5 levels, so the flip happens twice
t8 = TreeNode(1,
              TreeNode(2, TreeNode(4), TreeNode(5)),
              TreeNode(3, None,
                TreeNode(6, None,
                  TreeNode(7, None,
                    TreeNode(8)))))
print(sol.zigzagLevelOrder(t8))     # [[1], [3, 2], [4, 5, 6], [7], [8]]

# a chain - one node per level, so zigzag looks identical to level order
t5 = TreeNode(1, TreeNode(2, TreeNode(3, TreeNode(4, TreeNode(5)))))
print(sol.zigzagLevelOrder(t5))     # [[1], [2], [3], [4], [5]]
````
