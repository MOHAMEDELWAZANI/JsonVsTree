# JSON 03 - Nested data & missing keys  (where beginners crash)

Real JSON is nested: a dict holding a list of dicts, each holding another list.
And - the painful part - **not every record has every key**. Real data is dirty.

Your data: **`../data/orders.json`**

```json

  "shop": "TechMart",
  "currency": "MAD",
  "orders": [
    { "id": "A1",

      "items": [ { "product": "Laptop", "qty": 1, "price": 12000 }, ... ],
      "note": "gift wrap" },
    { "id": "A2",
      "customer": { "name": "Omar" },          <-- NO city
```

## Recap - 03

- Walk a chain one step at a time and know dict vs list at each step.
- `KeyError` = missing key. `.get(key, default)` when the key is genuinely optional.
- `.get("k", {})` to keep chaining safely.
- `in` on a dict tests keys.
